**Завдання 2**

− Написати процедуру генерації купи з будь-якого рандомного масива.
− Додати до нього елемент, який дорівнює вашому порядковому номеру у списку групи.
− Вилучити максимальний елемент з купи.
− Оцінити асимптотичну складність (у середньому і в найгіршомувипадку) процедур search, insert і delete роботи з купою.

```python
import random

heap = [0]
size = 0

def parent(i):
    return i // 2
def leftChild(i):
    return i * 2
def rightChild(i):
    return i * 2 + 1
def swapElements(i, j):
    heap[i], heap[j] = heap[j], heap[i]
# Додавання елемента в купу
def insertElement(value):
    global size

    heap.append(value)
    size += 1
    current = size
    while current > 1 and heap[parent(current)] < heap[current]:
        swapElements(current, parent(current))
        current = parent(current)
# Відновлення купи
def heapify(index):
    largest = index

    left = leftChild(index)
    right = rightChild(index)

    if left <= size and heap[left] > heap[largest]:
        largest = left

    if right <= size and heap[right] > heap[largest]:
        largest = right

    if largest != index:
        swapElements(index, largest)
        heapify(largest)
# Видалення максимального елемента
def removeMax():
    global size
    if size == 0:
        return None
    maxValue = heap[1]
    heap[1] = heap[size]
    heap.pop()
    size -= 1
    heapify(1)
    return maxValue

# Генерація випадкового масиву
numbers = []
for i in range(10):
    numbers.append(random.randint(1, 100))
print("Масив:", numbers)
# Створення купи
for num in numbers:
    insertElement(num)
print("Купа:", heap)
# Додавання номера у списку групи
insertElement(1)
print("Після додавання:", heap)
maximum = removeMax()
print("Максимальний елемент:", maximum)
print("Купа після видалення:", heap)
```

* Операція search. Для пошуку конкретного елемента може знадобитися перегляд усіх вузлів купи. O(n), O(n)
* Операція insert. Новий елемент додається в кінець купи, після чого піднімається вгору для відновлення властивості купи. O(logn), O(logn)
* Операція delete. Після видалення кореня останній елемент переноситься на його місце й опускається вниз. O(logn), O(logn)

– Проведіть тестування геш-таблиці з різними типами даних (цілі числа, рядки, списки, словники, об’єкти):
* перевірити працездатність геш-таблиці з різними типами даних.
* виміряти час виконання основних операцій (пошук, вставка, видалення) для різних типів даних.
* порівняти результати для різних типів даних.

```python
class HashTable:

    def __init__(self, size):
        self.size = size
        self.table = []
        for i in range(size):
            self.table.append([])

    def hash_function(self, key):
        return hash(str(key)) % self.size

    def insert(self, key, value):
        index = self.hash_function(key)
        self.table[index].append((key, value))

    def search(self, key):
        index = self.hash_function(key)
        bucket = self.table[index]
        for k, v in bucket:
            if k == key:
                return v
        return None

    def delete(self, key):

        index = self.hash_function(key)
        bucket = self.table[index]
        for i in range(len(bucket)):
            if bucket[i][0] == key:
                del bucket[i]
                return True
        return False
ht = HashTable(10)

ht.insert(5, "apple")
print(ht.search(5))

ht.insert("name", "test")
ht.insert(25, [7, 8, 9])
ht.insert((2, 3), {"x": 10})

print(ht.search("name"))
print(ht.search(25))
print(ht.search((2, 3)))

# видалення
ht.delete("name")

print(ht.search("name"))
```

Організоване як повне бінарне дерево в масиві.

**Додавання:** 

1) Додати елемент у кінець масиву (останній рівень дерева).

2) Порівняти його з батьківським вузлом.

3) Якщо порушено властивість купи:
* для max-heap: якщо дитина > батька → поміняти місцями
* для min-heap: якщо дитина < батька → поміняти місцями

4) Повторювати, поки елемент не стане на правильне місце або не досягне кореня.

**Вилучення елемента:**

1) Видаляється кореневий елемент.

2) Останній елемент переноситься у корінь.

3) Повторювати, поки властивість купи не відновиться.
